## **The original equations**

**The $z$ and $c$ equations are given by**
$$
z=\sqrt{\frac{\left(a^2-3\right)\left(3 a^2-1\right)}{5\left(a^2-1\right)}}, \quad c=\frac{10 a\left(a^2-1\right)^2}{\pi\left(9 a^4-30 a^2+9\right)}, \quad c_0=\frac{16 a\left(a^4-5 a^2+1\right)}{\pi\left(9 a^4-30 a^2+9\right)} .
$$

Using the condition that 
$$
\sqrt{c_0} + \sqrt{c_1} = |z|
$$
we get 

```python
%load_ext nbmagics
```

```python
%%wolfram

z[a_]  := Sqrt[((a^2 - 3) (3 a^2 - 1))/(5 (a^2 - 1))];
c[a_]  := (10 a (a^2 - 1)^2)/(Pi (9 a^4 - 30 a^2 + 9));
c0[a_] := (16 a (a^4 - 5 a^2 + 1))/(Pi (9 a^4 - 30 a^2 + 9));


NSolve[(c[a] + c0[a] - Abs[z[a]]^2)^2 == 4*c[a]*c0[a], a, Reals]
```

This gives the <span style="color:red">**critical value**</span> $3.5801189539434475$.

## **Conformal mapping equations**

Using the **conformal mapping** relations
$$
\begin{aligned}
z & =\lambda\left(\overline{x_0}+\frac{2 x}{1-|\lambda|^4}\right), \\
c & =x\left(\overline{x_0}+\frac{2 x\left(1+|\lambda|^4\right)}{\left(1-|\lambda|^4\right)^2}\right), \\
c_0 & =x_0\left(\overline{x_0}+2 x\right)
\end{aligned}
$$

we get the equation

$$
|z|^2\left(c_0-2 c\right)|\lambda|^{10}+\left[4\left(c_0-c\right)^2-|z|^4\right]|\lambda|^8-6 c_0|z|^2|\lambda|^6+2|z|^4|\lambda|^4-3|z|^2\left(c_0+2 c\right)|\lambda|^2+3|z|^4=0
$$

```python
%%wolfram

Pol5[a_, q_] :=
  Abs[z[a]]^2 (c0[a] - 2 c[a]) q^5
  + (4 (c0[a] - c[a])^2 - Abs[z[a]]^4) q^4
  - 6 c0[a] Abs[z[a]]^2 q^3
  + 2 Abs[z[a]]^4 q^2
  - 3 Abs[z[a]]^2 * (c0[a] + 2 c[a]) q
  + 3 Abs[z[a]]^4;

Pol5[a, q]
```

```python
%%wolfram
NSolve[Pol5[a, 0], a, Reals]
```

If we let $q = |\lambda|^2$, then
$$
P(q) =
|z|^2\left(c_0-2 c\right)q^{5}+\left[4\left(c_0-c\right)^2-|z|^4\right]q^4-6 c_0|z|^2q^3+2|z|^4q^2-3|z|^2\left(c_0+2 c\right)q+3|z|^4
$$
When $q = 0$, we get 
$$
P(0) = \frac{3 \left(3 a^4-10 a^2+3\right)^2}{25 \left(a^2-1\right)^2} > 0
$$
Also this value is $0$ when $a = \sqrt 3$.

## **For $1 < a < 1.1799822936820743$**

```python
%%wolfram

Clear[asol, qsol]

(* a < Sqrt[3] *)
nP5lo = FullSimplify[Pol5[a, q],  {1 < a < Sqrt[3]}];
nP4lo = FullSimplify[D[Pol5[a, q], q],  {1 < a < Sqrt[3]}];


{asol, qsol} = {a, q} /. NSolve[
    {nP4lo == 0, nP5lo == 0, q >= 0, a > 1, a < Sqrt[3]},
    {a, q},
    Reals    
][[1]] (* Note that there is only one solution *)

{
    D[nP4, q] /. {a -> asol, q -> qsol}, 
    D[nP5lo, a] /. {a -> asol, q -> qsol}
}


Plot[{
        Pol5[asol, q], 
        Pol5[asol+0.02, q]
    }, 
    {q, 0, 2},
    PlotLegends -> {
        "a=" <> ToString[asol], 
        "a=" <> ToString[asol + 0.02
        ]}
    ]  
```

```python
%%wolfram
ClearAll[confPhi, x0xb, x0xl, PlotConformalMaps];

aval = 2.18;
qvals = Module[{q},
  Sort[(q /. # &) /@ NSolve[{Pol5[aval, q] == 0, 0 < q < 1}, q, Reals]]
];

confPhi[w_, x0_, x_, lam_] :=
  w (Conjugate[x0] + 2 Conjugate[x]/(1 - Conjugate[lam]^2 w^2));

x0xb[a_, q_] := (1 - q^2)/(2 q^2 (3 + q^2)) *
  ((1 + q^2) (Abs[z[a]]^2/q - c0[a]) - 2 c[a]);

x0xl[a_, q_] := Module[{lam, x0, xl},
  lam = Sqrt[q];
  x0 = Sqrt[c0[a] - 2 x0xb[a, q]];
  xl = (c0[a] - x0^2)/(2 x0);
  {lam, x0, xl}
];

PlotConformalMaps[a_, q_] := Module[{lam, x0val, xval},
  {lam, x0val, xval} = x0xl[a, q];
  confPhi[#, x0val, xval, lam] & /@
    Table[Exp[2 Pi I t], {t, 0, 1, 0.001}]
];

If[qvals === {},
  "No real roots satisfy 0 < q < 1.",
  ComplexListPlot[
    (PlotConformalMaps[aval, #] &) /@ qvals,
    Joined -> True, PlotRange -> All
  ]
]
```

```python
%%wolfram
ClearAll[harmonicMoment];

harmonicMoment[a_, q_, k_Integer] := Module[
  {t, lam, x0, x, boundary, integrand},
  {lam, x0, x} = x0xl[a, q];
  boundary = confPhi[Exp[2 Pi I t], x0, x, lam];
  integrand = boundary^k Conjugate[boundary] D[boundary, t]/(2 I);
  Chop[NIntegrate[
    Evaluate[integrand], {t, 0, 1/4, 1/2, 3/4, 1},
    AccuracyGoal -> 9, PrecisionGoal -> 9
  ]]
];

If[qvals === {},
  "No real roots satisfy 0 < q < 1.",
  AssociationThread[{0, 1, 2, 4}, (harmonicMoment[aval, First[qvals], #] &) /@ {0, 1, 2, 4}]
]
```

```python
%%wolfram
If[Length[qvals] >= 2,
  AssociationThread[{0, 1, 2, 4}, (harmonicMoment[aval, qvals[[2]], #] &) /@ {0, 1, 2, 4}],
  "There is no second real root in 0 < q < 1."
]
```

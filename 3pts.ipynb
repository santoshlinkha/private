## **The original equations**

**The $z$ and $c$ equations are given by**
$$
z=\sqrt{\frac{\left(a^2-3\right)\left(3 a^2-1\right)}{5\left(a^2-1\right)}}, \quad c=\frac{10 a\left(a^2-1\right)^2}{\pi\left(9 a^4-30 a^2+9\right)}, \quad c_0=\frac{16 a\left(a^4-5 a^2+1\right)}{\pi\left(9 a^4-30 a^2+9\right)} .
$$

Using the condition that 
$$
\sqrt{c_0} + \sqrt{c_1} = |z|
$$
we get 

```python
%load_ext nbmagics
```

```python
%%wolfram

z[a_]  := Sqrt[((a^2 - 3) (3 a^2 - 1))/(5 (a^2 - 1))];
c[a_]  := (10 a (a^2 - 1)^2)/(Pi (9 a^4 - 30 a^2 + 9));
c0[a_] := (16 a (a^4 - 5 a^2 + 1))/(Pi (9 a^4 - 30 a^2 + 9));


NSolve[(c[a] + c0[a] - Abs[z[a]]^2)^2 == 4*c[a]*c0[a], a, Reals]
```

This gives the <span style="color:red">**critical value**</span> $3.5801189539434475$.

## **Conformal mapping equations**

Using the **conformal mapping** relations
$$
\begin{aligned}
z & =\lambda\left(\overline{x_0}+\frac{2 x}{1-|\lambda|^4}\right), \\
c & =x\left(\overline{x_0}+\frac{2 x\left(1+|\lambda|^4\right)}{\left(1-|\lambda|^4\right)^2}\right), \\
c_0 & =x_0\left(\overline{x_0}+2 x\right)
\end{aligned}
$$

we get the equation

$$
|z|^2\left(c_0-2 c\right)|\lambda|^{10}+\left[4\left(c_0-c\right)^2-|z|^4\right]|\lambda|^8-6 c_0|z|^2|\lambda|^6+2|z|^4|\lambda|^4-3|z|^2\left(c_0+2 c\right)|\lambda|^2+3|z|^4=0
$$

```python
%%wolfram

Pol5[a_, q_] :=
  Abs[z[a]]^2 (c0[a] - 2 c[a]) q^5
  + (4 (c0[a] - c[a])^2 - Abs[z[a]]^4) q^4
  - 6 c0[a] Abs[z[a]]^2 q^3
  + 2 Abs[z[a]]^4 q^2
  - 3 Abs[z[a]]^2 * (c0[a] + 2 c[a]) q
  + 3 Abs[z[a]]^4;

Pol5[a, q]
```

```python
%%wolfram
NSolve[Pol5[a, 0], a, Reals]
```

If we let $q = |\lambda|^2$, then
$$
P(q) =
|z|^2\left(c_0-2 c\right)q^{5}+\left[4\left(c_0-c\right)^2-|z|^4\right]q^4-6 c_0|z|^2q^3+2|z|^4q^2-3|z|^2\left(c_0+2 c\right)q+3|z|^4
$$
When $q = 0$, we get 
$$
P(0) = \frac{3 \left(3 a^4-10 a^2+3\right)^2}{25 \left(a^2-1\right)^2} > 0
$$
Also this value is $0$ when $a = \sqrt 3$.

```python
%%wolfram

NSolve[Numerator[FullSimplify[Pol5[a, 1], a>1]] == 0, a, Reals]
```

When $q = 1$, we get 
$$
P(1) = \frac{-80 \pi  (a-1) a (a+1) \left(1-3 a^2\right)^2 \left(a^2-3\right) \left(13 a^4-50 a^2+13\right) \left| a^2-3\right| +12 \pi ^2 \left(3 a^4-10 a^2+3\right)^4+1200 a^2 \left(a^6-11 a^4+11 a^2-1\right)^2}{75 \pi ^2 \left(-3 a^6+13 a^4-13 a^2+3\right)^2}
$$

Setting $P(1) = 0$, we get $a = 2.1898638923910614$ and $a = 3.5801189539434475$ which close to the root of $c_0 = 0 \implies a=2.188901059316734$. The other value is when the discs are tangent to each other.

## **For $1 < a < 1.1799822936820743$**

```python
%%wolfram

Clear[asol, qsol]

(* a < Sqrt[3] *)
nP5lo = FullSimplify[Pol5[a, q],  {1 < a < Sqrt[3]}];
nP4lo = FullSimplify[D[Pol5[a, q], q],  {1 < a < Sqrt[3]}];


{asol, qsol} = {a, q} /. NSolve[
    {nP4lo == 0, nP5lo == 0, q >= 0, a > 1, a < Sqrt[3]},
    {a, q},
    Reals    
][[1]] (* Note that there is only one solution *)

{
    D[nP4, q] /. {a -> asol, q -> qsol}, 
    D[nP5lo, a] /. {a -> asol, q -> qsol}
}


Plot[{
        Pol5[asol, q], 
        Pol5[asol+0.02, q]
    }, 
    {q, 0, 2},
    PlotLegends -> {
        "a=" <> ToString[asol], 
        "a=" <> ToString[asol + 0.02
        ]}
    ]  
```

Let $a_1 = 1.1799822936820743$, then from above, we conclude that $q > 1$ and has two solutions if $a< a_1$ and no solution if $a_1 < a < \sqrt 3$.

## **For $a > 2.1769602890498723$**

```python
%%wolfram

(* a > Sqrt[3] *)

nP5hi = FullSimplify[Pol5[a, q],  a > Sqrt[3]];
nP4hi = FullSimplify[D[Pol5[a, q], q],  a > Sqrt[3]];


{sol1, sol2} = {a, q} /. NSolve[
    {nP4hi == 0, nP5hi == 0, q >= 0,  a > Sqrt[3]},
    {a, q},
    Reals    
]

{
    D[nP4, q] /. {a -> sol1[[1]], q -> sol1[[2]]}, 
    D[nP5lo, a] /. {a -> sol1[[1]], q -> sol1[[2]]},
    D[nP4, q] /. {a -> sol2[[1]], q -> sol2[[2]]},
    D[nP5lo, a] /. {a -> sol2[[1]], q -> sol2[[2]]}
}
```

```python
%%wolfram
Plot[{
        Pol5[sol1[[1]], q], 
        Pol5[sol1[[1]]+0.02, q]
    }, 
    {q, 0, 2},
    PlotLegends -> {
        "a=" <> ToString[sol1[[1]]], 
        "a=" <> ToString[sol1[[1]] + 0.02
        ]}
    ]  
```

Let $a_2 = 2.1769602890498723$, then two real solutions merge at this value. Also other two real solution merge at $a^* = 3.5801189539434475$. Here, we notice that there are two valid real solution for $q$. This is valid until $P(1)  = 0$ which gives the value $a_3 = 2.1898638923910614$ and $a^* = 3.5801189539434475$ as calculated above. Also let $a_4 = 2.1941859956020435$, at this point the the largest root merges with the second largest one.

## **For $\sqrt 3 < a < a_2$**

We can see that the $q>1$ is the only solution as $\lim_{q \to \infty} P(q) = -\infty$.


## <span style="color:red">**Conclusion**</span>

|       Region      |            $q$           | 
|-------------------|--------------------------|
| 1 - $a_1$         | $q < 1$                  |
| $a_1$ - $\sqrt 3$ | No solution ($q < 0$)    |
| $\sqrt 3 - a_2$   | One solution ($q > 1$)   |
| $a_2$             | One solution $\|q\| < 1$ |
| $a_2$ - $a_3$     | 3 solutions (2 valid)    |
| $a_3$ - $a_4$     |  3 solutions (1 valid)   |
| $a_4$             |  2 solutions (1 valid)   |

a_1 &= 1.1799822936820743 \\
a_2 &= 2.1769602890498723 \\
a_3 &= 2.1898638923910614 \\

a^* &= 3.5801189539434475 \\
\end{aligned}
$$

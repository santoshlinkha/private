# **Three-point parameters**

For a given value of $a$, this notebook evaluates

$$
z=\sqrt{\frac{(a^2-3)(3a^2-1)}{5(a^2-1)}},\qquad
c=\frac{10a(a^2-1)^2}{\pi(9a^4-30a^2+9)},\qquad
c_0=\frac{16a(a^4-5a^2+1)}{\pi(9a^4-30a^2+9)}.
$$

## **Solve for $q=|\lambda|^2$**

With $q=|\lambda|^2$, the even tenth-degree equation becomes the quintic

$$
|z|^2(c_0-2c)q^5+[4(c_0-c)^2-|z|^4]q^4-6c_0|z|^2q^3
+2|z|^4q^2-3|z|^2(c_0+2c)q+3|z|^4=0.
$$

```matlab
a = 1.2;
z = sqrt(((a^2 - 3) * (3*a^2 - 1)) / (5 * (a^2 - 1)));
c = (10*a * (a^2 - 1)^2) / (pi * (9*a^4 - 30*a^2 + 9));
c0 = (16*a * (a^4 - 5*a^2 + 1)) / (pi * (9*a^4 - 30*a^2 + 9));

results = table(a, z, c, c0)

z2 = abs(z)^2;
z4 = z2^2;

q_coeff = [ ...
    z2*(c0 - 2*c), ...
    4*(c0 - c)^2 - z4, ...
    -6*c0*z2, ...
    2*z4, ...
    -3*z2*(c0 + 2*c), ...
    3*z4];

q = roots(q_coeff)
valid  = (abs(imag(q)) < 1e-10) & (real(q) > 0);
q = q(valid);

if ~isempty(q)
    residual = polyval(q_coeff, q);
    x0ConjX = ((1 - q.^2) ./ (2*q.^2.*(3 + q.^2))) .* ...
    ((1 + q.^2).*(abs(z)^2./q - c0) - 2*c);

absX0Squared = c0 - 2*x0ConjX;
x0 = sqrt(absX0Squared);
x = conj(x0ConjX ./ x0);

lambda = z.*(1 - q.^2) ./ ...
    (x0.*(1 - q.^2) + 2*x);


lambda4 = abs(lambda).^4;

zComputed = lambda .* ...
    (conj(x0) + 2*x ./ (1 - lambda4));

cComputed = x .* ...
    (conj(x0) + 2*x.*(1 + lambda4) ./ (1 - lambda4).^2);

c0Computed = x0 .* (conj(x0) + 2*x);

zResidual  = abs(zComputed - z);
cResidual  = abs(cComputed - c);
c0Residual = abs(c0Computed - c0);

results = table( ...
    lambda, x0, x, ...
    zComputed, cComputed, c0Computed, ...
    zResidual, cResidual, c0Residual, ...
    'VariableNames', { ...
    'lambda', 'x0', 'x', ...
    'zComputed', 'cComputed', 'c0Computed', ...
    'zResidual', 'cResidual', 'c0Residual' ...
    })
    

else
    disp('No valid roots found for q.')
end

```
